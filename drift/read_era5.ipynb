```python
new_df = pd.read_csv("era5_october_data.csv", index_col=0, parse_dates=True)
```

```python
new_df
```

```python
target_lat, target_lon = 40.0, -75.0

coords = np.array(new_df[['latitude', 'longitude']].values)
target = np.array([[target_lat, target_lon]])

distances = cdist(target, coords, metric='euclidean')
nearest_idx = distances.argmin(axis=1)
closest_row = new_df.iloc[nearest_idx[0]]
closest_row
```

```python
wind_x, wind_y = closest_row['u'], closest_row['v']
wind_x, wind_y
```
